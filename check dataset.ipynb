```python

import pandas as pd
import os

print("HEALTHCARE RECOMMENDATION PROJECT")
print("--------------------------------")

file_path = "dataset/Original_Dataset.csv"

if os.path.exists(file_path):
    data = pd.read_csv(file_path)

    print("\nDataset loaded successfully!")
    print("\nFirst 5 rows:")
    print(data.head())

    print("\nDataset Shape:")
    print(data.shape)

    print("\nColumn Names:")
    print(data.columns.tolist())

    print("\nMissing Values:")
    print(data.isnull().sum())

else:
    print("Dataset file not found. Check the folder location.")
```
